```python
import pandas as pd

BASE = "produtos_regioes.csv"
df = pd.read_csv(BASE)
df
```

```python
pd.crosstab(df["regiao"], df["produto"])
```

```python
pd.crosstab(df['regiao'], df['produto'], margins=True,margins_name='Vendas')
```

```python
pd.crosstab(df['regiao'], df['produto'],normalize=True, margins=True, margins_name='Total')
```

```python
pd.crosstab(
    df['regiao'], df['produto'],normalize=True, margins=True, margins_name='Total'
    ).mul(100)
```

```python
pd.crosstab(df['regiao'], df['produto'], 
            margins=True,margins_name='Vendas')
```

```python
pd.crosstab(df['regiao'], df['produto'],normalize="index", 
            margins=True,margins_name='Total').mul(100).round(1)
```

```python
pd.crosstab(
    df['regiao'], df['produto'],normalize="columns", 
            margins=True,margins_name='Total').mul(100).round(1)
```

```python
pd.crosstab(df['regiao'], df['produto'])
```

```python
pd.crosstab(df['regiao'], df['produto'], values=df['receita'], 
            aggfunc='sum', margins=True, margins_name='Receita Total').round(2)
```

```python
pd.crosstab(df['regiao'], df['produto'], values=df['receita'], 
            aggfunc='sum', normalize=True).mul(100).round(2)
```

```python
pd.crosstab(df['regiao'], df['produto'], values=df['receita'], 
            aggfunc='sum', normalize="index").mul(100).round(2)
```

```python
pd.crosstab(df['regiao'], df['produto'], values=df['receita'], 
            aggfunc='sum', normalize='columns').mul(100).round(2)
```

```python
pd.crosstab(df['regiao'], df['produto'])
```

```python
df.groupby(['regiao', 'produto']).count().unstack().fillna(0).rename(columns={'receita':'vendas'})
```
