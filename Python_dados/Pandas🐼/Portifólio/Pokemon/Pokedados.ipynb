```python

# from google.colab import drive
# drive.mount('/content/drive')


import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, HTML
```

```python
BASE = 'pokemon.csv'

df_pokedex = pd.read_csv(BASE)

df_pokedex.head()
```

```python
df_pokedex.info()
```

```python
df_pokedex['Type 2'] = df_pokedex['Type 2'].fillna('Nulo')

df_pokedex
```

```python
df_pokedex = df_pokedex.rename(columns={'Name':'Nome','Type 1':'Tipo 1','Type 2':'Tipo 2','Attack':'Ataque','Defense':'Defesa','Sp. Atk':'Ataque Especial','Sp. Def':'Defesa Especial','Speed':'Velocidade','Generation':'Geracao','Legendary':'Lendario'})
```

```python
sorted(df_pokedex['Tipo 1'].unique())
```

```python
sorted(df_pokedex['Tipo 2'].unique())
```

```python
df_pokedex['Tipo 1'] = df_pokedex['Tipo 1'].replace({
'Bug':'Inseto',
'Dark':'Sombrio',
'Dragon':'Dragao',
'Electric':'Eletrico',
'Fairy':'Fada',
'Fighting':'Lutador',
'Fire':'Fogo',
'Flying':'Voador',
'Ghost':'Fantasma',
'Grass':'Planta',
'Ground':'Terra',
'Ice':'Gelo',
'Poison':'Venenoso',
'Psychic':'Psiquico',
'Rock':'Pedra',
'Steel':'Aco',
'Water':'Agua'})
```

```python
df_pokedex['Tipo 2'] = df_pokedex['Tipo 2'].replace({
'Bug':'Inseto',
'Dark':'Sombrio',
'Dragon':'Dragao',
'Electric':'Eletrico','Fairy':'Fada',
'Fighting':'Lutador',
'Fire':'Fogo',
'Flying':'Voador',
'Ghost':'Fantasma',
'Grass':'Planta',
'Ground':'Terra',
'Ice':'Gelo',
'Poison':'Venenoso',
'Psychic':'Psiquico',
'Rock':'Pedra',
'Steel':'Aco',
'Water':'Agua'})
```

```python
df_pokedex
```

```python
df_pokemon_kanto_jhoto = df_pokedex[df_pokedex['Geracao'] <= 2].copy()

df_pokemon_kanto_jhoto
```

```python
df_pokemon_kanto_jhoto.info()
```

```python
df_pokemon_kanto_jhoto['Lendario'] = df_pokemon_kanto_jhoto['Lendario'].astype(str)
```

```python
df_pokemon_kanto_jhoto['Lendario'] = df_pokemon_kanto_jhoto['Lendario'].replace({
'True':'Sim',
'False':'Não'})
```

```python
#Aqui já terminei de modelar o CSV original e deixei do
#jeito que eu queria

df_pokemon_kanto_jhoto.style.hide(axis='index')
df_pokemon_kanto_jhoto
```

```python
df_media_ataque_tipo1 = df_pokemon_kanto_jhoto.groupby('Tipo 1')['Ataque'].mean(numeric_only=True).reset_index().round(2)
df_media_ataque_tipo1
```

```python
plt.figure(figsize=(20,10))

sns.barplot(data= df_media_ataque_tipo1,
            x='Tipo 1',
            y='Ataque',
            hue='Tipo 1',
            errorbar=None)
plt.title("Média Ataques Pokemons Tipo 1")
plt.xlabel("Ataques por tipos")
plt.ylabel("Média dos Ataques")
```

```python
df_media_ataque_tipo2 = df_pokemon_kanto_jhoto.groupby('Tipo 2')['Ataque'].mean(numeric_only=True).reset_index().round(2)
df_media_ataque_tipo2
```

```python
plt.figure(figsize=(20,10))

sns.barplot(data= df_media_ataque_tipo2,
            x='Tipo 2',
            y='Ataque',
            hue='Tipo 2',
            errorbar=None)
plt.title("Média Ataques Pokemons Tipo 2")
plt.xlabel("Ataques por tipos")
plt.ylabel("Média dos Ataques")
```

```python
#Selecionando os tipos 1 por grupamento

df_grupamento_1 = df_pokemon_kanto_jhoto.groupby('Tipo 1')[['Ataque','Defesa','Velocidade']].mean(numeric_only=True).reset_index().round(2)
df_grupamento_1
df_melted_grupamento_1 = df_grupamento_1.melt(
    id_vars='Tipo 1',
    value_vars=['Ataque','Defesa','Velocidade'],
    var_name='Atributos',
    value_name='Média'
)

df_melted_grupamento_1
```

```python
plt.figure(figsize=(15,8))

sns.barplot(data=df_melted_grupamento_1,
            x='Tipo 1',
            y='Média',
            hue='Atributos')

plt.yticks(np.arange(0, 220, 20))

plt.title("Média de Ataque, Defesa e Velocidade")
plt.xlabel('Primeiro Tipo dos Pokemons')
plt.ylabel('Status')
plt.show()
```

```python
#Aqui é a mediana entre os atributos que selecionei acima
#separando eles por tipo de Pokémon

df_mediana = df_pokemon_kanto_jhoto.groupby('Tipo 1')[['Ataque','Defesa','Velocidade']].median(numeric_only=True).reset_index().round(2)
df_melt_mediana = df_mediana.melt(
    id_vars='Tipo 1',
    value_vars=['Ataque','Defesa','Velocidade'],
    var_name='Atributos',
    value_name='Mediana'
)
df_melt_mediana
```

```python
plt.figure(figsize=(15,8))

sns.barplot(data=df_melt_mediana,
            x='Tipo 1',
            y='Mediana',
            hue='Atributos')
plt.yticks(np.arange(0,210,20))
plt.title("Mediana do 1º Tipo de Pokemons")
plt.xlabel("Tipos")
plt.ylabel('Status')
plt.show()
```

```python
#Aqui tô fazendo a média entre ataque e defesa para ver quais são
#os tipos de pokémons mais eficientes
df_eficiencia = df_pokemon_kanto_jhoto.groupby('Tipo 1')[['Ataque','Defesa']].mean(numeric_only=True).reset_index().round(2)

df_eficiencia['Eficiencia_ofensiva'] = df_eficiencia['Ataque']/df_eficiencia['Defesa']

df_eficiencia.sort_values(by='Eficiencia_ofensiva', ascending=False).round(2)
```

```python
#Aqui eu peguei os pokémons com maiores valores da média
#entre defesa e HP.

df_outliers = df_pokemon_kanto_jhoto.groupby('Nome')[['Defesa','HP']].mean(numeric_only=True).round(2).reset_index()
df_media_defesa = df_pokemon_kanto_jhoto['Defesa'].mean(numeric_only=True)


#Aqui é um cálculo pra saber quanto o maior "tanque"
#tá acima da defesa geral entre os maiores defensores

df_outliers.sort_values(by=['Defesa','HP'], ascending=[False, False]).head(5)
df_media_defesa.round(2)
df_acima_media = df_outliers['Defesa'].head(1)/df_media_defesa

display(
    df_outliers.sort_values(by=['Defesa','HP'], ascending=[False, False]).head(5),
    df_media_defesa.round(2),
    df_acima_media
)
```

```python
df_media_defesa = df_pokemon_kanto_jhoto.groupby('Tipo 1').agg({
'Ataque':'mean',
'Defesa':'mean'})

df_media_defesa.round(2).rename(columns={
'Ataque':'Ataque Médio',
'Defesa':'Defesa Média'}).sort_values(by=['Ataque Médio','Defesa Média'],ascending=[False,False])

#O grupo aço tem a melhor quantidade de ataque
#e defesa
```

```python
df_agg = df_pokemon_kanto_jhoto.groupby('Tipo 1').agg({
'Ataque':'mean',
'Defesa':'max',
'Nome':'count'})

df_agg.round(2).rename(columns={
'Ataque':'Ataque Médio',
'Defesa':'Defesa Média',
'Nome':'Total Pokémon'}).sort_values(by='Ataque Médio',ascending=False)
```

```python
#Procurando os 'tanques' de combate

df_tanques = df_pokemon_kanto_jhoto[(df_pokemon_kanto_jhoto['Defesa Especial'] >= 100) & (df_pokemon_kanto_jhoto['HP'] >= 80)]
df_5_maiores = df_tanques[['Nome','HP','Defesa Especial']].sort_values(by=['Defesa Especial','HP'],ascending=[False, False]).head(5)

df_melted = df_5_maiores.melt(
    id_vars=['Nome'],
    value_vars=['HP','Defesa Especial'],
    var_name='Atributos',
    value_name='Valor'
)
df_melted
```

```python
plt.Figure(figsize=(25,8))
sns.barplot(data=df_melted,
            x='Valor',
            y='Nome',
            hue='Atributos',
            palette=['#4C72B0', '#55A868'])
plt.title("5 Tanques: Comparação de HP e Defesa")
plt.xlabel('Pokémon')
plt.ylabel('Atributos')

plt.show()
```

```python
#A razão dos três pokémon mais ofensivo

df_ataque_defesa = df_pokemon_kanto_jhoto['Razão Ataque-Defesa']=df_pokemon_kanto_jhoto['Ataque']/df_pokemon_kanto_jhoto['Defesa']
df_ataque_defesa = df_pokemon_kanto_jhoto[['Nome','Razão Ataque-Defesa']].sort_values(by='Razão Ataque-Defesa', ascending=False).round(2).head(3)
df_ataque_defesa

#Os resultados não mostram que necessariamente eles sejam os mais fortes para lutar
```

```python
#Os mais velozes

df_velocidade = df_pokemon_kanto_jhoto.groupby(['Geracao','Tipo 1']).agg({
'Velocidade':['mean','max']}).sort_values(by=('Velocidade','mean'), ascending=False).head(5).round(2).reset_index()
df_velocidade
```

```python
plt.figure(figsize=(10,8))

sns.barplot(data=df_velocidade_melted,
            x='Tipo 1',
            y='Valor',
            hue='Metrica')
plt.title('Os tipos mais rápidos')
plt.xlabel('Tipos de Pokemon')
plt.ylabel('Velocidade')
plt.show()
```

```python
#Mew x Mewtwo

df_mew = df_pokemon_kanto_jhoto[(df_pokemon_kanto_jhoto['Nome'] == 'Mew') | (df_pokemon_kanto_jhoto['Nome'] == 'Mewtwo')]
df_mew.sort_values(by='Total', ascending=False).round(2)

#Os dados mostram que (infelizmente) o Mewtwo supera o Mew
```

```python
df_melt_mew = df_mew.melt(
    id_vars='Nome',
    value_vars=['Total','Ataque','Defesa','Ataque Especial','Defesa Especial','Velocidade'],
    var_name='Atributos',
    value_name='Status'
)

df_melt_mew
```

```python
plt.figure(figsize=(10,8))

sns.barplot(data=df_melt_mew,
            x='Nome',
            y='Status',
            hue='Atributos')
plt.yticks(np.arange(0,710, 30))
plt.title("Mew x Mewtwo")
plt.xlabel("Embate Direto")
plt.ylabel("Status")
plt.show()
```

```python
#Agora incluindo o Dragonite...

df_dragonite = df_pokemon_kanto_jhoto[(df_pokemon_kanto_jhoto['Nome'] == 'Mew') | (df_pokemon_kanto_jhoto['Nome'] == 'Mewtwo') | (df_pokemon_kanto_jhoto['Nome'] == 'Dragonite')]
df_dragonite.sort_values(by='Total', ascending=False).round(2)

#O Dragonite é algum bug desse jogo. Certeza!
```

```python
plt.figure(figsize=(10,8))

sns.barplot(data=df_melt_dragonite,
            x='Nome',
            y='Status',
            hue='Atributos')
plt.yticks(np.arange(0,710, 30))
plt.title("Dragonite entre Mew x Mewtwo")
plt.xlabel("Embate Direto")
plt.ylabel("Status")
plt.show()
```
