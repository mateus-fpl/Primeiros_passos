```python
import pandas as pd

BASE = "dados_satisfacao.csv"

dados = pd.read_csv(BASE)

dados
```

```python
dados.info()
```

```python
dados_melt = pd.melt(
    dados,
    id_vars=["cliente_id","idade","genero"],
    value_vars=[
        "satisfacao_geral_01-2024",
        "satisfacao_geral_02-2024",
        "satisfacao_geral_03-2024",
        "satisfacao_produto_01-2024",
        "satisfacao_produto_02-2024",
        "satisfacao_produto_03-2024",
        "satisfacao_servico_01-2024",
        "satisfacao_servico_02-2024",
        "satisfacao_servico_03-2024",
    ],
    var_name="categoria_mes",
    value_name="nota"
)

dados_melt
```

```python
dados_melt.head(25)
```

```python
dados_melt.info()
```

```python
dados_melt[["categoria","mês"]] = dados_melt["categoria_mes"].str.rsplit("_",n=1, expand=True)
```

```python
dados_melt
dados_melt.info()
```

```python
dados_melt['mês'] = pd.to_datetime(dados_melt["mês"], format="%m-%Y")

dados_melt
```

```python
dados_melt.info()
```

```python
dados_melt["mês"] = dados_melt["mês"].dt.month
dados_melt
```

```python
dados_melt["genero"] = dados_melt["genero"].astype("category")
dados_melt["categoria"] = dados_melt["categoria"].astype("category")
dados_melt
```
