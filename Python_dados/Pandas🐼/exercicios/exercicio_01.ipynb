```python
# Nova Coluna / Condicional: Crie uma coluna nova chamada 'Urgente'.

# Se o tempo for maior que 40 minutos, o valor deve ser True.

# Se for menor ou igual a 40, o valor deve ser False.
# (Dica: dá pra fazer direto com df['Tempo_Minutos'] > 40)
```

```python
import pandas as pd

dados = {
    'ID_Exame': [101, 102, 103, 104, 105, 106, 107],
    'Paciente': ['Ana', 'Bruno', 'Carlos', 'Daniela', 'Eduardo', 'Fernanda', 'Gabriel'],
    'Setor': ['Hematologia', 'Bioquimica', 'Hematologia', 'Imunologia', 'Bioquimica', 'Hematologia', 'Imunologia'],
    'Tempo_Minutos': [12, 45, 18, 60, 30, 15, 90],
    'Status': ['Concluido', 'Pendente', 'Concluido', 'Pendente', 'Concluido', 'Concluido', 'Cancelado']
}

df_relatorio = pd.DataFrame(dados)
df_relatorio
```

```python
df_visao_geral = ["Paciente","Setor","Status"]
df_relatorio[df_visao_geral]
```

```python
df_hemato = df_relatorio[df_relatorio["Setor"]=="Hematologia"]
df_hemato
```

```python
df_problemas = df_relatorio[(df_relatorio["Status"]== "Pendente") & (df_relatorio["Tempo_Minutos"] >30)]
df_problemas
```

```python
df_relatorio.groupby("Setor")["Tempo_Minutos"].mean().reset_index()
```

```python
df_relatorio["Urgente"] = df_relatorio["Tempo_Minutos"] > 40
df_relatorio
```
