```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
```

```python
x,y = make_regression(n_samples=500, n_features=1,noise=10)
plt.scatter(x, y, color='blue', alpha=0.3)
plt.title("Nossa Base de Dados (500 Amostras)")
plt.xlabel("Input (X)")
plt.ylabel("Target (Y)")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

modelo = LinearRegression()

modelo.fit(x, y)

previsoes = modelo.predict(x)

# 4. Mostrar o resultado final
plt.figure(figsize=(10, 6))
plt.scatter(x, y, color='blue', alpha=0.3, label='Dados Reais (com ruído)')
plt.plot(x, previsoes, color='red', linewidth=3, label='Previsão da IA (Tendência)')
plt.title("O Sucesso do Modelo: Linha de Regressão Linear")
plt.legend()
plt.show()
```

```python
nota = modelo.score(x, y)

print(f"A precisão do modelo (R²) é: {nota:.2f}")
```
